# MODELS

```python
from paddleocr import PPStructureV3

pipeline = PPStructureV3(
    text_recognition_model_name="PP-OCRv5_server_rec",
    use_doc_orientation_classify=False, # Use use_doc_orientation_classify to enable/disable document orientation classification model
    use_doc_unwarping=False,    # Use use_doc_unwarping to enable/disable document unwarping module
    use_textline_orientation=False, # Use use_textline_orientation to enable/disable textline orientation classification model
    device="gpu:0", # Use device to specify GPU for model inference
)
```

# CONVERSION

```python
import os

PDF_DIR = f"{DATA_DIR}/pdf"
PARSED_DOCUMENTS_DIR = f"{ROOT_DIR}/parser_documents/paddleocr"

os.makedirs(PARSED_DOCUMENTS_DIR, exist_ok=True)
```

```python
import re
from pydantic import BaseModel
from typing import List, Optional, Union


class Coordinates(BaseModel):
    x0: Union[int, float]
    y0: Union[int, float]
    x1: Union[int, float]
    y1: Union[int, float]


class LayoutElement(BaseModel):
    text: str
    id: Union[str, int]
    classification: Optional[str] = None
    coordinates: Optional[Coordinates] = None


class ParsedPage(BaseModel):
    elements: List[LayoutElement] = list()


def convert_to_markdown_coords_only(paddle_output):
    page = ParsedPage()

    for i, obj in enumerate(paddle_output["parsing_res_list"]):
        coords = obj.bbox
        clean_text = obj.content

        if not clean_text:
            continue
        if not coords:
            continue

        if coords and len(coords)==4:
            page.elements.append(LayoutElement(
                **{
                    "text": clean_text,
                    "id": i,
                    "coordinates": Coordinates(
                        **{
                            "x0": coords[0] / 1000,
                            "y0": coords[1] / 1000,
                            "x1": coords[2] / 1000,
                            "y1": coords[3] / 1000,
                        }
                    )
                }
            ))
        elif coords and len(coords)==8:
            coords = list(map(int, coords))
            page.elements.append(LayoutElement(
                **{
                    "text": clean_text,
                    "id": i,
                    "coordinates": Coordinates(
                        **{
                            "x0": min(coords[::2]) / 1000,
                            "y0": min(coords[1::2]) / 1000,
                            "x1": max(coords[::2]) / 1000,
                            "y1": max(coords[1::2]) / 1000,
                        }
                    )
                }
            ))
        else:
            page.elements.append(LayoutElement(
                **{
                    "text": clean_text,
                    "id": i,
                    "coordinates": None
                }
            ))

    return page
```

```python
import glob

# already processed documents
paths = glob.glob(f"{PARSED_DOCUMENTS_DIR.replace('paddleocr', 'granit_docling')}/**/*.json", recursive=True)
proc_docs = set([x.split("/")[-1].replace(".json", "") for x in paths])
```

```python
import glob
import fitz
import json
import base64
import io
from PIL import Image
from io import BytesIO
from tqdm.auto import tqdm
import numpy as np

pdf_paths = glob.glob(f"{PDF_DIR}/*.pdf")
pdf_paths = list(filter(
    lambda x: x.split("/")[-1].replace(".pdf", "") in proc_docs,
    pdf_paths
))
target_size = 1024

for path in tqdm(pdf_paths):
    filename = path.split("/")[-1].replace(".pdf", "")
    out_path = f"{PARSED_DOCUMENTS_DIR}/{filename}.json"
    if os.path.exists(out_path):
        continue

    doc = fitz.open(path)

    outputs = list()
    for page_num in range(2, len(doc)):
        page = doc.load_page(page_num)
        pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))
        img_data = pix.tobytes("png")
        img = Image.open(io.BytesIO(img_data))
        img.thumbnail((target_size, target_size))

        output = pipeline.predict(
            input=np.array(img)
        )

        outputs.append(output[0])


    parsed_doc = dict()
    for page_n, output in enumerate(outputs, 1):
        parsed_doc[page_n] = convert_to_markdown_coords_only(output).model_dump()

    # ---- save ----

    with open(out_path, "w") as f:
        json.dump(parsed_doc, f)
```
